````python
!pip install -q langchain-groq langchain-core

import os
import re
import json
from getpass import getpass

from langchain_groq import ChatGroq
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

# --------------------------------
# 1. API KEY (FIXED)
# --------------------------------
# Fix: the old error "Illegal header value b'Bearer gsk_... '" was caused by a
# trailing space/newline in the pasted key. We now strip all whitespace.
# Preferred: add a Colab Secret named GROQ_API_KEY (🔑 icon in the left sidebar).

def clean_key(key):
    """Remove spaces, newlines, tabs and quotes accidentally copied with the key."""
    return re.sub(r"\s+", "", key or "").strip("'\"")

try:
    from google.colab import userdata
    api_key = clean_key(userdata.get("GROQ_API_KEY"))
except Exception:
    api_key = clean_key(os.environ.get("GROQ_API_KEY"))

if not api_key:
    api_key = clean_key(getpass("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 "))

os.environ["GROQ_API_KEY"] = api_key

# --------------------------------
# 2. INITIALIZE LLM
# --------------------------------

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.7
)

# --------------------------------
# 3. LONG-TERM MEMORY
# --------------------------------
# In Colab this file is lost when the runtime resets.
# To keep it, mount Google Drive and point memory_file to a path inside it.

memory_file = "user_memory.json"

if os.path.exists(memory_file):
    with open(memory_file, "r") as file:
        user_memory = json.load(file)
else:
    user_memory = {}

# --------------------------------
# 4. CONVERSATION HISTORY
# --------------------------------

chat_history = []

# --------------------------------
# 5. HELPER: PARSE JSON SAFELY
# --------------------------------

def parse_json(text):
    """Parse JSON even if the model wraps it in ```json fences."""
    text = re.sub(r"^```(?:json)?|```$", "", text.strip(), flags=re.MULTILINE).strip()
    return json.loads(text)

# --------------------------------
# 6. CHAT FUNCTION
# --------------------------------

def chat(user_input):

    global user_memory

    system_prompt = f"""
    You are a professional AI assistant.

    User information stored in memory:
    {json.dumps(user_memory)}

    Instructions:
    1. Remember information shared by the user.
    2. If the user tells you their name, identify it.
    3. Use the user's name when relevant.
    4. Answer naturally and professionally.
    5. Never invent personal information.
    """

    messages = [SystemMessage(content=system_prompt)]
    messages.extend(chat_history)
    messages.append(HumanMessage(content=user_input))

    response = llm.invoke(messages)

    # --------------------------------
    # 7. EXTRACT USER INFORMATION
    # --------------------------------

    extraction_prompt = f"""
    Analyze the following user message.

    Current memory:
    {json.dumps(user_memory)}

    User message:
    {user_input}

    Extract any personal information explicitly shared by the user.

    Return ONLY valid JSON.

    Example:
    {{"name": "Dharani"}}

    If no new personal information is shared, return:
    {{}}

    Do not guess or invent information.
    """

    extraction_response = llm.invoke(extraction_prompt)

    try:
        new_memory = parse_json(extraction_response.content)
        user_memory.update(new_memory)

        with open(memory_file, "w") as file:
            json.dump(user_memory, file, indent=4)

    except (json.JSONDecodeError, AttributeError, TypeError, ValueError):
        print("(Memory extraction failed.)")

    # --------------------------------
    # 8. SAVE CHAT HISTORY
    # --------------------------------

    chat_history.append(HumanMessage(content=user_input))
    chat_history.append(AIMessage(content=response.content))

    return response.content


# --------------------------------
# 9. INTERACTIVE CHATBOT
# --------------------------------

print("AI Assistant Started!")
print("Commands: 'memory' shows saved info, 'forget' clears it, 'exit' stops.\n")

while True:

    user_input = input("You: ").strip()

    if not user_input:
        continue

    if user_input.lower() == "exit":
        print("AI: Goodbye!")
        break

    if user_input.lower() == "memory":
        print("Saved memory:", json.dumps(user_memory, indent=2), "\n")
        continue

    if user_input.lower() == "forget":
        user_memory = {}
        chat_history.clear()
        if os.path.exists(memory_file):
            os.remove(memory_file)
        print("AI: Memory cleared.\n")
        continue

    try:
        answer = chat(user_input)
        print("\nAI:", answer)
        print()
    except Exception as e:
        print(f"\n(Error: {e})\n")
````
