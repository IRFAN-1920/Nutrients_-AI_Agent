````python
# ============================================
# 1. INSTALL REQUIRED PACKAGES
# ============================================

!pip install -q groq "mcp[cli]>=1.28,<2"

# ============================================
# 2. IMPORT LIBRARIES
# ============================================

import os
import re
import json
import time
import sys
import asyncio
import subprocess
from getpass import getpass
from groq import Groq
from mcp import ClientSession
from mcp.client.streamable_http import streamablehttp_client

# ============================================
# 3. GROQ API KEY
# ============================================
# Preferred: Colab Secret named GROQ_API_KEY (🔑 icon in the left sidebar).
# Falls back to a hidden prompt. Whitespace is stripped to avoid
# the "Illegal header value" error.

def clean_key(key):
    return re.sub(r"\s+", "", key or "").strip("'\"")

try:
    from google.colab import userdata
    api_key = clean_key(userdata.get("GROQ_API_KEY"))
except Exception:
    api_key = clean_key(os.environ.get("GROQ_API_KEY"))

if not api_key:
    api_key = clean_key(getpass("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 "))

os.environ["GROQ_API_KEY"] = api_key

# ============================================
# 4. CREATE MCP SERVER FILE
# ============================================

server_code = r'''
import os
import re
import json
from groq import Groq
from mcp.server.fastmcp import FastMCP

client = Groq(api_key=os.environ["GROQ_API_KEY"])

MODEL = "openai/gpt-oss-120b"

mcp = FastMCP("Nutrition AI Agent")


def parse_json(text):
    text = re.sub(r"^```(?:json)?|```$", "", text.strip(), flags=re.MULTILINE).strip()
    return json.loads(text)


@mcp.tool()
def analyze_food(food: str) -> str:
    """
    Analyze a food item and return exactly 6 important
    nutritional details.
    """

    prompt = f"""
    You are a professional nutrition analysis assistant.

    Analyze the following food: {food}

    Assume a serving size of 100 grams unless the food
    is normally measured differently.

    Return valid JSON with exactly these 7 fields
    (the food name plus 6 nutritional details):

    {{
        "food": "Food name and serving size",
        "calories": "Calories in kcal",
        "protein": "Protein in grams",
        "carbohydrates": "Carbohydrates in grams",
        "fat": "Fat in grams",
        "vitamins_minerals": "Important vitamins and minerals",
        "health_benefits": "Health benefits"
    }}

    Rules:
    - Always use the same 7 JSON fields shown above.
    - Give approximate nutritional values.
    - Specify whether the food is raw or cooked when relevant.
    - Do not invent exact values when reliable data is unavailable.
    - Keep the health benefits concise.
    - Return JSON only.
    """

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "user", "content": prompt}
        ],
        response_format={"type": "json_object"},
        temperature=0.2
    )

    data = parse_json(response.choices[0].message.content)

    def fmt(value):
        if isinstance(value, list):
            return ", ".join(str(v) for v in value)
        return value

    output = f"""
NUTRITION ANALYSIS

Food: {fmt(data.get('food', food))}

1. Calories:
   {fmt(data.get('calories', 'Not available'))}

2. Protein:
   {fmt(data.get('protein', 'Not available'))}

3. Carbohydrates:
   {fmt(data.get('carbohydrates', 'Not available'))}

4. Fat:
   {fmt(data.get('fat', 'Not available'))}

5. Vitamins & Minerals:
   {fmt(data.get('vitamins_minerals', 'Not available'))}

6. Health Benefits:
   {fmt(data.get('health_benefits', 'Not available'))}
"""

    return output


if __name__ == "__main__":
    mcp.run(transport="streamable-http")
'''

with open("nutrition_server.py", "w") as file:
    file.write(server_code)

print("MCP Server file created successfully!")


# ============================================
# 5. START MCP SERVER
# ============================================

# Stop an older server started in this notebook, if any.
try:
    if server_process.poll() is None:
        server_process.terminate()
        server_process.wait(timeout=5)
except (NameError, Exception):
    pass

# Errors go to a log file so you can debug if the server fails to start.
server_log = open("server.log", "w")

server_process = subprocess.Popen(
    [sys.executable, "nutrition_server.py"],
    stdout=server_log,
    stderr=server_log,
    env=os.environ.copy()
)

time.sleep(5)

if server_process.poll() is not None:
    print(open("server.log").read())
    raise RuntimeError("MCP Server failed to start. See the log above.")

print("MCP Server started successfully!")


# ============================================
# 6. CONNECT MCP CLIENT
# ============================================

async def analyze_with_mcp(food):

    async with streamablehttp_client(
        "http://127.0.0.1:8000/mcp"
    ) as (read, write, _):

        async with ClientSession(read, write) as session:

            await session.initialize()

            result = await session.call_tool(
                "analyze_food",
                arguments={
                    "food": food
                }
            )

            if result.isError:
                print("Error:", result.content)
                return

            for content in result.content:
                if content.type == "text":
                    print(content.text)


# ============================================
# 7. GET USER INPUT
# ============================================

food_name = input("Enter a food name: ")

if food_name.strip():
    await analyze_with_mcp(food_name.strip())
else:
    print("Please enter a valid food name.")
````
